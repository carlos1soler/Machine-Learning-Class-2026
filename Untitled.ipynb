```python
# ==============================
# CONFIG
# ==============================
NEWS_API_KEY = "dc5a3253f4bc42a59c3db08e473d985f"

# ==============================
# STEP 1: Get worst 15 days
# ==============================
worst_50_days = worst_50_days.nsmallest(50, 'returns').copy()

# ==============================
# STEP 2: Function to fetch news
# ==============================
def get_news_and_summary(row):
    date = row['DATE']
    date_str = date.strftime('%Y-%m-%d')

    url = "https://newsapi.org/v2/everything"
    params = {
        "q": "oil OR crude OR WTI OR OPEC",
        "from": date_str,
        "to": date_str,
        "sortBy": "relevancy",
        "language": "en",
        "apiKey": NEWS_API_KEY,
        "pageSize": 5
    }

    try:
        response = requests.get(url, params=params)
        data = response.json()

        if data.get("articles"):
            headlines = [a['title'] for a in data['articles']]
            headlines_text = " | ".join(headlines)

            # Simple grounded summary (NO hallucination)
            summary = headlines[0]  # take most relevant headline

            return pd.Series([headlines_text, summary])

        else:
            return pd.Series(["No relevant news", "No clear event"])

    except Exception as e:
        return pd.Series(["Error fetching news", "Error"])

# ==============================
# STEP 3: Apply
# ==============================
worst_15_days[['headlines', 'summary']] = worst_15_days.apply(get_news_and_summary, axis=1)

# ==============================
# STEP 4: Final output
# ==============================
cols = ['DATE', 'returns', 'Per_change', 'headlines', 'summary']
result = worst_15_days[cols]

print(result)
```
